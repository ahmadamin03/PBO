```python
# Tugas Object & Class
# Nama: (Isi nama Anda)
# NIM: (Isi NIM Anda)

# Definisikan class
class Mahasiswa:
    def __init__(self, nama, nim):
        self.nama = nama
        self.nim = nim

    def tampilkan_data(self):
        print("Nama:", self.nama)
        print("NIM:", self.nim)

# Membuat object dari class Mahasiswa
mhs1 = Mahasiswa("Budi", "12345")
mhs2 = Mahasiswa("Ani", "67890")

# Menampilkan data mahasiswa
print("Data Mahasiswa 1:")
mhs1.tampilkan_data()

print("\nData Mahasiswa 2:")
mhs2.tampilkan_data()
```
